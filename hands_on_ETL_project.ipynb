Step 1 — Handle Missing Values

Cell 4



```python
# Fill missing department
df["department"] = df["department"].fillna("Unknown")

# Convert salary to numeric first
df["salary"] = (
    df["salary"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

# Fill missing salary with median salary
median_salary = df["salary"].median()

df["salary"] = df["salary"].fillna(median_salary)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

Step 2 — Remove Duplicates
Cell 5

```python
print("Number of duplicate rows before:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nNumber of duplicate rows after:")
print(df.duplicated().sum())
```

Step 3 — Standardize Department Names

Cell 6

```python
df["department"] = (
    df["department"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print(df["department"].value_counts())
```

Step 4 — Clean Salary

Cell 7

```python
print(df["salary"].head())

print("\nSalary data type:")
print(df["salary"].dtype)
```

Step 5 — Create Bonus

Cell 8

```python
df["bonus"] = df["salary"] * 0.10

print(df[["employee_name", "salary", "bonus"]])
```

Step 6 — Create Total Salary

Cell 9

```python
df["total_salary"] = df["salary"] + df["bonus"]

print(df[[
    "employee_name",
    "salary",
    "bonus",
    "total_salary"
]])
```

Step 7 — Final Clean Data

Cell 10

```python
print("Final Clean Data")
display(df)
```

Step 9 — Create Database Table

Cell 15

```python
import sqlite3

# 1. Create a Database connection in Google Colab
connection = sqlite3.connect('my_database.db')

# 2. Put your code right below this
cursor = connection.cursor()

create_table_query = """
CREATE TABLE IF NOT EXISTS employees (
    employee_id INT,
    employee_name VARCHAR(100),
    department VARCHAR(100),
    salary DECIMAL(10,2),
    bonus DECIMAL(10,2),
    total_salary DECIMAL(10,2)
)
"""

cursor.execute(create_table_query)
print("Table created successfully!")
```

Step 10 — Insert Data into MySQL

Cell 16

```python
insert_query = """
INSERT INTO employees
(employee_id, employee_name, department, salary, bonus, total_salary)
VALUES (?, ?, ?, ?, ?, ?)
"""

for _, row in df.iterrows():
    values = (
        int(row["employee_id"]),
        row["employee_name"],
        row["department"],
        float(row["salary"]),
        float(row["bonus"]),
        float(row["total_salary"])
    )

    cursor.execute(insert_query, values)

connection.commit()
print("Data inserted successfully!")
```

Step 11 — SQL Analysis

Query 1 — Highest Salary by Department

```python
query1 = """
SELECT
    department,
    MAX(salary) AS highest_salary
FROM employees
GROUP BY department
ORDER BY highest_salary DESC;
"""

cursor.execute(query1)

results = cursor.fetchall()

for row in results:
    print(row)
```

Query 2 — Average Salary

```python
query2 = """
SELECT
    AVG(salary) AS average_salary
FROM employees;
"""

cursor.execute(query2)

result = cursor.fetchone()

print("Average Salary:", result[0])
```

Query 3 — Number of Employees by Department

```python
query3 = """
SELECT
    department,
    COUNT(*) AS number_of_employees
FROM employees
GROUP BY department
ORDER BY number_of_employees DESC;
"""

cursor.execute(query3)

results = cursor.fetchall()

for row in results:
    print(row)
```
